# String & Slice 

```python
str = "Aayush is good boy"
str1 = "monty Python"
print(str[0:6])
print(len(str))
print(str[0:6:2])
print(str[-12:-9])
print(str1[-12:-7])
print(str1[::-1])
print(str.endswith("boy"))
print(str.count("o"))
print(str1.capitalize())
print(str.lower()) # same as upper
print(str1.replace("monty","Aayush"))
print(str1.find("t"))
print(str1.find("TT"))## give -1 when not found 
```



 ### SYNTAX : [including:excluding:skipChar]

    - including is Starting Point And Excluding is ending point of string
    - set parameeter it is set by default 0 [:6:2] it means [0:6:2]
    - if skip_char not given it set by default 1 give skip_char > len(str) then it try possible resuolve
    
 **Negative Parameter**
  
     - string start from 0,1... so on and if you want reverse string then giv it Negative Parameter 
     - in above example -1 is "y" so including is -4 and exculding is -2 
  
  
      - give Skipchar in Negative Parameter it reverse string
